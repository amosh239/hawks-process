```python
from src import load_and_prep_data, SeasonalPoisson, HawkesExp
from src import temporal_train_test_split, calc_test_ll, evaluate_forecast
```

```python
# Load
sequences = load_and_prep_data('data/raw/retail_data.csv')
user_id = sequences.apply(len).idxmax() 
user_times = sequences[user_id]
train, test = temporal_train_test_split(user_times, 0.8)
```

```python
# Fit
poisson = SeasonalPoisson().fit(train)
hawkes = HawkesExp(baseline_model=poisson).fit(train)
```

```python
# Evaluate
ll_p = calc_test_ll(poisson, train, test)
ll_h = calc_test_ll(hawkes, train, test)
print(f"LL/event -> Poisson: {ll_p:.4f} | Hawkes: {ll_h:.4f}")

rmse_p, _, _ = evaluate_forecast(poisson, train, test)
rmse_h, _, _ = evaluate_forecast(hawkes, train, test)
print(f"RMSE (h) -> Poisson: {rmse_p:.2f} | Hawkes: {rmse_h:.2f}")
```
